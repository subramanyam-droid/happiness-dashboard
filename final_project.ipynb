```python
import pandas as pd
import plotly.express as px

df = pd.read_csv("2019.csv")
df.head()
```

```python
df.columns = df.columns.str.strip()
df = df.dropna()
print(df.columns.tolist())
```

## Question 1: How does GDP per capita relate to happiness score, and does that relationship differ by region?

```python
import plotly.express as px

fig = px.scatter(
    df,
    x="GDP per capita",
    y="Score",
    color="Country or region",  # <-- Changed from "Region"
    color_discrete_sequence=px.colors.qualitative.Safe,
    title="Wealth buys happiness — but the payoff varies sharply by region",
)
fig.show()
```

**Answer:** There's a clear positive relationship between GDP per capita and happiness overall, but the strength of that link differs by region. Western European and North American countries reach high happiness scores even at moderate GDP levels, while some wealthy Middle Eastern nations sit lower on happiness than their GDP alone would predict — showing wealth is necessary but not sufficient for happiness.

## Q4 / Healthy Life Expectancy vs. Happiness by Region


```python
fig = px.scatter(
    df, 
    x="Healthy life expectancy",  # Corrected from "Life Expectancy"
    y="Score",                    # Corrected from "Happiness Score"
    color="Region",
    color_discrete_sequence=px.colors.qualitative.Safe,
    title="Longer, healthier lives line up with higher happiness across regions",
    template="plotly_white"
)
fig.show()
```

## Q6 / Freedom vs. Generosity

```python
coeffs = np.polyfit(df["GDP per capita"], df["Score"], 1)
df["Predicted"] = coeffs[0] * df["GDP per capita"] + coeffs[1]
df["Residual"] = df["Score"] - df["Predicted"]

top_over = df.nlargest(5, "Residual")
top_under = df.nsmallest(5, "Residual")
combined = pd.concat([top_over, top_under])

fig = px.bar(
    combined.sort_values("Residual"), 
    x="Residual", 
    y="Country or region",  # Corrected from "Country"
    orientation="h",
    color="Residual", 
    color_continuous_scale="RdBu",
    title="Costa Rica and similar nations punch above their GDP weight in happiness",
    template="plotly_white"
)
fig.show()
```

## Q8 / Top 10 vs. Bottom 10 Countries by Happiness

```python
top10 = df.nlargest(10, "Score")
bottom10 = df.nsmallest(10, "Score")
combined = pd.concat([top10, bottom10])

fig = px.bar(
    combined.sort_values("Score"), 
    x="Score", 
    y="Country or region",  # Corrected from "Country"
    orientation="h",
    color="Region", 
    color_discrete_sequence=px.colors.qualitative.Safe,
    title="Nordic countries dominate the top 10; conflict-affected nations dominate the bottom",
    template="plotly_white"
)
fig.show()
```

## Q9 / Distribution of Key Factors Across Countries

```python
melted = df.melt(
    id_vars="Region", 
    value_vars=["GDP per capita", "Social support", "Freedom to make life choices", "Generosity"],
    var_name="Factor", 
    value_name="Value"
)

fig = px.box(
    melted, 
    x="Factor", 
    y="Value", 
    color="Factor",
    color_discrete_sequence=px.colors.qualitative.Safe,
    title="Generosity varies far more unpredictably across countries than GDP does",
    template="plotly_white"
)
fig.show()
```

## Q10 / Radar Chart for Top 5 Happiest Countries

```python
top5 = df.nlargest(5, "Score")
categories = [
    "GDP per capita", 
    "Social support", 
    "Healthy life expectancy", 
    "Freedom to make life choices", 
    "Generosity"
]

fig = px.line_polar(
    top5.melt(id_vars="Country or region", value_vars=categories, var_name="Factor", value_name="Value"),
    r="Value", 
    theta="Factor", 
    color="Country or region",  # Corrected from "Country"
    line_close=True,
    title="The top 5 happiest countries share strong scores across every factor",
    template="plotly_white"
)
fig.show()
```
